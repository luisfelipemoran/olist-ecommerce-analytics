# Metodología del Notebook 5 - Preparación de tablas finales y exportes

## Objetivo

El objetivo de este notebook es organizar y exportar los principales datasets agregados que servirán como base para el dashboard final y para la comunicación de resultados del proyecto. Esta fase no busca generar nuevos hallazgos, sino estructurar la información ya analizada en formatos claros, consistentes y reutilizables.

## Metodología

En este notebook se preparan las tablas finales del proyecto para su uso en visualización y comunicación de resultados. A partir de la base enriquecida con variables de negocio y sentimiento, se construyen indicadores agregados y tablas resumen orientadas al dashboard y a la presentación ejecutiva.

## Procedimiento

A partir de la base maestra enriquecida con variables de negocio, logística, geografía, categorías y sentimiento, se construyen tablas resumen con distintos niveles de agregación. Estas tablas incluyen KPIs generales del ecommerce, evolución temporal de pedidos y ventas, métricas logísticas, resultados por categoría, resultados por estado y distribución del sentimiento en las reseñas.

Posteriormente, cada tabla se exporta en formato CSV para facilitar su utilización en herramientas de visualización como Power BI, Tableau o similares. De este modo, el notebook actúa como puente entre la fase analítica y la fase de presentación ejecutiva del proyecto.

## Resultado esperado

El resultado final de este notebook es un conjunto de archivos limpios y estructurados, preparados para alimentar el dashboard y facilitar la elaboración de la memoria técnica y la presentación final.

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 200)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
```

```python
df = pd.read_csv("../data/final/master_orders_with_sentiment.csv")
df.head()
```

```python
df.shape
```

### Aseguramos el formato correcto de las fechas:

```python
date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
    "review_creation_date",
    "review_answer_timestamp"
]

for col in date_cols:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")
```

### Creamos variables temporales para el dashboard

```python
df["purchase_year"] = df["order_purchase_timestamp"].dt.year
df["purchase_month"] = df["order_purchase_timestamp"].dt.month
df["purchase_month_name"] = df["order_purchase_timestamp"].dt.month_name()
df["purchase_year_month"] = df["order_purchase_timestamp"].dt.to_period("M").astype(str)

df[["order_purchase_timestamp", "purchase_year", "purchase_month", "purchase_year_month"]].head()
```

### Definición de los KPIs

```python
kpis_dashboard = pd.DataFrame({
    "KPI": [
        "Total pedidos",
        "Clientes únicos",
        "Ventas totales (R$)",
        "Ticket promedio (R$)",
        "Review score promedio",
        "% pedidos tardíos",
        "% sentimiento positivo",
        "% sentimiento negativo",
        "% sentimiento neutro"
    ],
    "Valor": [
        len(df),
        df["customer_unique_id"].nunique(),
        round(df["payment_value_total"].sum(), 2),
        round(df["payment_value_total"].mean(), 2),
        round(df["review_score"].mean(), 2),
        round(df["is_late"].mean() * 100, 2),
        round((df["sentiment_3"] == "Positivo").mean() * 100, 2),
        round((df["sentiment_3"] == "Negativo").mean() * 100, 2),
        round((df["sentiment_3"] == "Neutro").mean() * 100, 2)
    ]
}).reset_index(drop=True)

kpis_dashboard
```

### Evolución mensual

```python
monthly_dashboard = (
    df
    .groupby("purchase_year_month", as_index=False)
    .agg(
        pedidos=("order_id", "count"),
        ventas=("payment_value_total", "sum"),
        ticket_promedio=("payment_value_total", "mean"),
        review_score_promedio=("review_score", "mean"),
        porcentaje_tardios=("is_late", "mean")
    )
)

monthly_dashboard["ventas"] = monthly_dashboard["ventas"].round(2)
monthly_dashboard["ticket_promedio"] = monthly_dashboard["ticket_promedio"].round(2)
monthly_dashboard["review_score_promedio"] = monthly_dashboard["review_score_promedio"].round(2)
monthly_dashboard["porcentaje_tardios"] = (monthly_dashboard["porcentaje_tardios"] * 100).round(2)

monthly_dashboard = monthly_dashboard.sort_values("purchase_year_month").reset_index(drop=True)
monthly_dashboard.head()
```

### Creación de categorias

```python
category_df = df[df["main_category"].notna()].copy()
category_df = category_df[category_df["main_category"] != "Unknown"]

category_dashboard = (
    category_df
    .groupby("main_category", as_index=False)
    .agg(
        pedidos=("order_id", "count"),
        ventas=("payment_value_total", "sum"),
        ticket_promedio=("payment_value_total", "mean"),
        review_score_promedio=("review_score", "mean"),
        porcentaje_tardios=("is_late", "mean")
    )
)

category_dashboard["ventas"] = category_dashboard["ventas"].round(2)
category_dashboard["ticket_promedio"] = category_dashboard["ticket_promedio"].round(2)
category_dashboard["review_score_promedio"] = category_dashboard["review_score_promedio"].round(2)
category_dashboard["porcentaje_tardios"] = (category_dashboard["porcentaje_tardios"] * 100).round(2)

category_dashboard = category_dashboard.sort_values("ventas", ascending=False).reset_index(drop=True)
category_dashboard.head(10)
```

### Distribución Geografica

```python
state_dashboard = (
    df
    .groupby("geolocation_state", as_index=False)
    .agg(
        pedidos=("order_id", "count"),
        ventas=("payment_value_total", "sum"),
        ticket_promedio=("payment_value_total", "mean"),
        review_score_promedio=("review_score", "mean"),
        porcentaje_tardios=("is_late", "mean")
    )
)

state_dashboard["ventas"] = state_dashboard["ventas"].round(2)
state_dashboard["ticket_promedio"] = state_dashboard["ticket_promedio"].round(2)
state_dashboard["review_score_promedio"] = state_dashboard["review_score_promedio"].round(2)
state_dashboard["porcentaje_tardios"] = (state_dashboard["porcentaje_tardios"] * 100).round(2)

state_dashboard = state_dashboard.sort_values("ventas", ascending=False).reset_index(drop=True)
state_dashboard.head(10)
```

## Analisis logistico

```python
delivered_df = df[df["order_status"] == "delivered"].copy()

logistics_dashboard = pd.DataFrame({
    "KPI": [
        "Pedidos entregados",
        "Tiempo entrega promedio (días)",
        "Mediana entrega (días)",
        "Retraso promedio (días)",
        "% pedidos tardíos entregados"
    ],
    "Valor": [
        len(delivered_df),
        round(delivered_df["delivery_days"].mean(), 2),
        round(delivered_df["delivery_days"].median(), 2),
        round(delivered_df["delay_days"].mean(), 2),
        round(delivered_df["is_late"].mean() * 100, 2)
    ]
})

logistics_dashboard
```

## Analisis de sentimiento

```python
sentiment_dashboard = (
    df[df["sentiment_3"].notna()]["sentiment_3"]
    .value_counts()
    .rename_axis("Sentimiento")
    .reset_index(name="Cantidad")
)

sentiment_dashboard["Porcentaje"] = (
    sentiment_dashboard["Cantidad"] / sentiment_dashboard["Cantidad"].sum() * 100
).round(2)

sentiment_dashboard = sentiment_dashboard.reset_index(drop=True)
sentiment_dashboard
```

## Sentimientos por categoria

```python
sentiment_category_df = df[
    df["main_category"].notna() &
    (df["main_category"] != "Unknown") &
    df["sentiment_3"].notna()
].copy()

sentiment_category_dashboard = (
    sentiment_category_df
    .groupby("main_category", as_index=False)["sentiment_3"]
    .value_counts(normalize=True)
)

sentiment_category_dashboard["proportion"] = (
    sentiment_category_dashboard["proportion"] * 100
).round(2)

sentiment_category_dashboard = sentiment_category_dashboard.reset_index(drop=True)
sentiment_category_dashboard.head(20)
```

## Sentimiendo por geolocalizacion:

```python
sentiment_category_pivot_dashboard = sentiment_category_dashboard.pivot(
    index="main_category",
    columns="sentiment_3",
    values="proportion"
).fillna(0).reset_index()

sentiment_category_pivot_dashboard.head(10)
```

## Sentimientos por Estados

```python
sentiment_state_df = df[
    df["geolocation_state"].notna() &
    df["sentiment_3"].notna()
].copy()

sentiment_state_dashboard = (
    sentiment_state_df
    .groupby("geolocation_state", as_index=False)["sentiment_3"]
    .value_counts(normalize=True)
)

sentiment_state_dashboard["proportion"] = (
    sentiment_state_dashboard["proportion"] * 100
).round(2)

sentiment_state_dashboard = sentiment_state_dashboard.reset_index(drop=True)
sentiment_state_dashboard.head(20)
```

```python
sentiment_state_pivot_dashboard = sentiment_state_dashboard.pivot(
    index="geolocation_state",
    columns="sentiment_3",
    values="proportion"
).fillna(0).reset_index()

sentiment_state_pivot_dashboard.head(20)
```

## Clientes existentes con indicadores clave

```python
customer_dashboard = (
    df
    .groupby("customer_unique_id", as_index=False)
    .agg(
        total_pedidos=("order_id", "count"),
        gasto_total=("payment_value_total", "sum"),
        ticket_promedio=("payment_value_total", "mean"),
        review_score_promedio=("review_score", "mean")
    )
)

customer_dashboard["segmento_frecuencia"] = np.where(
    customer_dashboard["total_pedidos"] == 1,
    "Compra única",
    "Cliente repetidor"
)

customer_dashboard["gasto_total"] = customer_dashboard["gasto_total"].round(2)
customer_dashboard["ticket_promedio"] = customer_dashboard["ticket_promedio"].round(2)
customer_dashboard["review_score_promedio"] = customer_dashboard["review_score_promedio"].round(2)

customer_dashboard.head()
```

## Resumen de clientes para el dashboard

```python
customer_segment_dashboard = (
    customer_dashboard
    .groupby("segmento_frecuencia", as_index=False)
    .agg(
        clientes=("customer_unique_id", "count"),
        gasto_total_promedio=("gasto_total", "mean"),
        ticket_promedio_cliente=("ticket_promedio", "mean"),
        review_score_promedio=("review_score_promedio", "mean")
    )
)

customer_segment_dashboard["gasto_total_promedio"] = customer_segment_dashboard["gasto_total_promedio"].round(2)
customer_segment_dashboard["ticket_promedio_cliente"] = customer_segment_dashboard["ticket_promedio_cliente"].round(2)
customer_segment_dashboard["review_score_promedio"] = customer_segment_dashboard["review_score_promedio"].round(2)

customer_segment_dashboard = customer_segment_dashboard.reset_index(drop=True)
customer_segment_dashboard
```

```python
kpis_dashboard.to_csv("../outputs/dashboard/kpis_dashboard.csv", index=False)
monthly_dashboard.to_csv("../outputs/dashboard/monthly_dashboard.csv", index=False)
category_dashboard.to_csv("../outputs/dashboard/category_dashboard.csv", index=False)
state_dashboard.to_csv("../outputs/dashboard/state_dashboard.csv", index=False)
logistics_dashboard.to_csv("../outputs/dashboard/logistics_dashboard.csv", index=False)
sentiment_dashboard.to_csv("../outputs/dashboard/sentiment_dashboard.csv", index=False)
customer_dashboard.to_csv("../outputs/dashboard/customer_dashboard.csv", index=False)
customer_segment_dashboard.to_csv("../outputs/dashboard/customer_segment_dashboard.csv", index=False)

sentiment_category_pivot_dashboard.to_csv("../outputs/dashboard/sentiment_category_pivot_dashboard.csv", index=False)
sentiment_state_pivot_dashboard.to_csv("../outputs/dashboard/sentiment_state_pivot_dashboard.csv", index=False)
```

```python
from pathlib import Path

out = Path("outputs/dashboard")
out.mkdir(parents=True, exist_ok=True)
```

```python
def exportar_tablas_looker(tablas, carpeta_salida):
    exportadas = []

    for nombre, df in tablas.items():
        archivo = carpeta_salida / f"{nombre}.xlsx"
        df.to_excel(archivo, index=False)
        exportadas.append(archivo.name)

    return exportadas
```

```python
tablas_looker = {
    "01_kpis_dashboard": kpis_dashboard,
    "02_monthly_dashboard": monthly_dashboard,
    "03_category_dashboard": category_dashboard,
    "04_state_dashboard": state_dashboard,
    "05_customer_segment_dashboard": customer_segment_dashboard,
    "06_sentiment_dashboard": sentiment_dashboard,
    "07_sentiment_category_dashboard": sentiment_category_dashboard,
    "08_sentiment_state_dashboard": sentiment_state_dashboard
}
```

```python
archivos_exportados = exportar_tablas_looker(tablas_looker, out)
archivos_exportados
```
